It only works for one constraint at a time. In the appropriate cell below, we have to enter the da oops constraint num, and start and end date of the se flows we want to read

```python
def get_panorama_constraint_uids(da_oops_con_num):
    SPP_ID = download_df_from_sql_db(f"select ConstraintName from spp_constraints.ConstraintNameAndDetails_v3 where constraintNum = {da_oops_con_num}").iloc[0,0]    
    (SPP_ID)
    con_details = all_spp_constraints[all_spp_constraints["iso_constraint_labels"].str.contains(SPP_ID)].to_dict(orient='records')[0]
    # print(con_details)
    return con_details['monitored_uid'], con_details['contingency_uid']
```

```python
def get_outage_decomposition(mon_uid, con_uid, start_dt, end_dt):
    path_ = f"/spp-se/constraint/outage_decomposition"
    payload = {"collection" : "spp-se",
              "monitored_uid" : mon_uid,
              "contingency_uid" : con_uid,
              "start_datetime" : start_dt,
              "end_datetime" : end_dt}
    return _get_panorama_data("reflow", path_, requests.get, payload)
```

```python
def plot_outage_decomposition(df):
    df["newlegend"] = df["hypercluster_uid"].str.replace(",",",\n").apply(lambda x:x[:35])
    fig = px.area(df, x="timestamp", y="impact", color="newlegend", line_group="newlegend")    
    fig.show()


outage_decomp = get_outage_decomposition(mon_uid, con_uid,start_dte, end_dte)
plot_outage_decomposition(outage_decomp)
```

```python
def get_constraint_flow(mon_uid, con_uid, start_dt, end_dt):
    path_ = f"/spp-se/constraint/flow"
    payload = {"collection" : "spp-se",
              "monitored_uid" : mon_uid,
              "contingency_uid" : con_uid,
              "start_datetime" : start_dt,
              "end_datetime" : end_dt}
    return _get_panorama_data("reflow", path_, requests.get, payload)
flow = get_constraint_flow(mon_uid, con_uid, start_dte, end_dte)
display(flow.head(2))
fig = px.area(flow, x="timestamp", y="constraint_flow")    
fig.show()
```

```python
def get_driver_decomp(mon_uid, con_uid, start_dt, end_dt, driver_type):
    path_ = f"/spp-se/constraint/driver_decomposition"
    payload = {"collection" : "spp-se",
              "monitored_uid" : mon_uid,
              "contingency_uid" : con_uid,
              "start_datetime" : start_dt,
              "end_datetime" : end_dt,
              "driver_type" : driver_type}
    return _get_panorama_data("reflow", path_, requests.get, payload)

#Get gen, zonal load and fuel decomposition here
gen_decomp = get_driver_decomp(mon_uid, con_uid, start_dte, end_dte,"GENERATION")
display(gen_decomp.head(10))
load_decomp = get_driver_decomp(mon_uid, con_uid, start_dte, end_dte,"LOAD_ZONE")
display(load_decomp.head(2))
fuel_decomp = get_driver_decomp(mon_uid, con_uid, start_dte, end_dte,"FUEL")
display(fuel_decomp.head(2))
```

```python
#This cell displays gen impact by year month

gen_decomp['mth'] = pd.to_datetime(gen_decomp['timestamp'],utc=True).dt.month
gen_decomp['yr'] = pd.to_datetime(gen_decomp['timestamp'],utc=True).dt.year

display(gen_decomp.head(2))

df_gen = gen_decomp.groupby(['yr','mth','driver_uid']).mean('impact').sort_values(['yr','mth']).reset_index()
display(df_gen.head(2))
df_gen.pivot(index = 'driver_uid', columns = ['yr','mth'], values = 'impact').round(0)
```
